```python
# Run this in a terminal (not Jupyter) or use: !python snake_game.py
# For Jupyter, here's the game logic demo:

import random
from collections import deque

class SnakeGame:
    def __init__(self, rows=10, cols=10):
        self.rows, self.cols = rows, cols
        self.snake = deque([(5, 5), (5, 4), (5, 3)])
        self.direction = (0, 1)
        self.food = self._place_food()
        self.score = 0
        self.alive = True

    def _place_food(self):
        empty = [(r,c) for r in range(self.rows) for c in range(self.cols)
                 if (r,c) not in self.snake]
        return random.choice(empty) if empty else None

    def move(self, direction=None):
        if direction: self.direction = direction
        head = (self.snake[0][0]+self.direction[0],
                self.snake[0][1]+self.direction[1])
        if (head[0]<0 or head[0]>=self.rows or head[1]<0 or
            head[1]>=self.cols or head in self.snake):
            self.alive = False
            return
        self.snake.appendleft(head)
        if head == self.food:
            self.score += 10
            self.food = self._place_food()
        else:
            self.snake.pop()

    def display(self):
        for r in range(self.rows):
            row = ""
            for c in range(self.cols):
                if (r,c) == self.snake[0]: row += "🐍"
                elif (r,c) in self.snake: row += "🟩"
                elif (r,c) == self.food: row += "🍎"
                else: row += "⬛"
            print(row)
        print(f"Score: {self.score}")

# Demo: simulate 5 moves
game = SnakeGame()
moves = [(0,1),(0,1),(1,0),(1,0),(0,1)]
for m in moves:
    game.move(m)
    if not game.alive: break
game.display()
```
